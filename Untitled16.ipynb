```python
import os
import numpy as np
import pandas as pd
import librosa
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from tqdm.auto import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from einops import rearrange, repeat
from einops.layers.torch import Rearrange

# Configuration
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🔧 Device: {DEVICE}")

# Paramètres
SAMPLE_RATE = 16000
DURATION = 5
MAX_LENGTH = SAMPLE_RATE * DURATION

# Paramètres spectrogramme
N_MELS = 128
N_FFT = 2048
HOP_LENGTH = 512

BATCH_SIZE = 8 if torch.cuda.is_available() else 4
EPOCHS = 8
LEARNING_RATE = 1e-4

BASE_PATH = '/content/drive/MyDrive/dev_data_ToyConveyor/ToyConveyor'
```

```python
from google.colab import drive
drive.mount('/content/drive')

class ToyConveyorAudioDataset(Dataset):
    """Dataset avec extraction de Mel-Spectrogramme"""

    def __init__(self, file_list, label_list, sample_rate=16000, n_mels=128,
                 duration=5, augment=False):
        self.files = file_list
        self.labels = label_list
        self.sample_rate = sample_rate
        self.n_mels = n_mels
        self.max_length = sample_rate * duration
        self.augment = augment

    def __len__(self):
        return len(self.files)

    def load_and_process_audio(self, file_path):
        try:
            # Charger audio
            audio, _ = librosa.load(file_path, sr=self.sample_rate, duration=5, mono=True)

            # Padding/truncate
            if len(audio) > self.max_length:
                audio = audio[:self.max_length]
            else:
                audio = np.pad(audio, (0, max(0, self.max_length - len(audio))))

            # Normalisation
            audio = audio / (np.max(np.abs(audio)) + 1e-8)

            # Mel-Spectrogramme
            mel_spec = librosa.feature.melspectrogram(
                y=audio,
                sr=self.sample_rate,
                n_mels=self.n_mels,
                n_fft=2048,
                hop_length=512
            )

            # Conversion en échelle log
            mel_spec_db = librosa.power_to_db(mel_spec, ref=np.max)

            # Normalisation
            mel_spec_db = (mel_spec_db - mel_spec_db.mean()) / (mel_spec_db.std() + 1e-8)

            return mel_spec_db.astype(np.float32)

        except Exception as e:
            print(f"⚠️ Erreur: {os.path.basename(file_path)}")
            return np.zeros((self.n_mels, 157), dtype=np.float32)

    def __getitem__(self, idx):
        mel_spec = self.load_and_process_audio(self.files[idx])
        label = self.labels[idx]

        # Augmentation (SpecAugment léger)
        if self.augment and np.random.random() > 0.5:
            # Time masking
            if np.random.random() > 0.5:
                t_mask_len = np.random.randint(1, 10)
                t_mask_start = np.random.randint(0, mel_spec.shape[1] - t_mask_len)
                mel_spec[:, t_mask_start:t_mask_start + t_mask_len] = 0

            # Frequency masking
            if np.random.random() > 0.5:
                f_mask_len = np.random.randint(1, 10)
                f_mask_start = np.random.randint(0, mel_spec.shape[0] - f_mask_len)
                mel_spec[f_mask_start:f_mask_start + f_mask_len, :] = 0

        return torch.FloatTensor(mel_spec).unsqueeze(0), torch.LongTensor([label])[0]


def load_dataset(base_path):
    """Charge le dataset"""
    print("🔍 Chargement du dataset...")

    all_files = []
    all_labels = []

    for root, dirs, files in os.walk(base_path):
        for file in files:
            if file.endswith('.wav'):
                full_path = os.path.join(root, file)
                all_files.append(full_path)

                path_str = full_path.lower()
                if 'anomaly' in path_str or 'abnormal' in path_str:
                    all_labels.append(1)
                else:
                    all_labels.append(0)

    print(f"✅ {len(all_files)} fichiers trouvés")
    print(f"   Normal: {all_labels.count(0)}")
    print(f"   Anomaly: {all_labels.count(1)}")

    # Split
    train_files, test_files, train_labels, test_labels = train_test_split(
        all_files, all_labels, test_size=0.2, random_state=42, stratify=all_labels
    )

    return train_files, test_files, train_labels, test_labels


# Charger les données
train_files, test_files, train_labels, test_labels = load_dataset(BASE_PATH)

train_dataset = ToyConveyorAudioDataset(train_files, train_labels, augment=True)
test_dataset = ToyConveyorAudioDataset(test_files, test_labels, augment=False)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

print(f"\n✅ Dataloaders créés:")
print(f"   Train batches: {len(train_loader)}")
print(f"   Test batches: {len(test_loader)}")

# Test
sample_spec, sample_label = train_dataset[0]
print(f"\n📊 Shape mel-spectrogramme: {sample_spec.shape}")
```

```python
class PatchEmbedding(nn.Module):
    """Découpe le spectrogramme en patches comme ViT"""
    def __init__(self, n_mels=128, time_steps=157, patch_size=16, embed_dim=256):
        super().__init__()

        self.patch_size = patch_size
        self.n_patches_h = n_mels // patch_size
        self.n_patches_w = time_steps // patch_size
        self.n_patches = self.n_patches_h * self.n_patches_w

        # Projection des patches
        self.projection = nn.Conv2d(1, embed_dim, kernel_size=patch_size, stride=patch_size)

    def forward(self, x):
        # x: (batch, 1, n_mels, time_steps)
        x = self.projection(x)  # (batch, embed_dim, H, W)
        x = rearrange(x, 'b e h w -> b (h w) e')
        return x


class AudioTransformer(nn.Module):
    """Transformer pour classification audio (inspiré de ViT)"""

    def __init__(self, n_mels=128, time_steps=157, patch_size=16,
                 embed_dim=256, depth=6, num_heads=8, mlp_ratio=4,
                 num_classes=2, dropout=0.1):
        super().__init__()

        # Patch embedding
        self.patch_embed = PatchEmbedding(n_mels, time_steps, patch_size, embed_dim)
        n_patches = self.patch_embed.n_patches

        # CLS token
        self.cls_token = nn.Parameter(torch.randn(1, 1, embed_dim))

        # Position embedding
        self.pos_embed = nn.Parameter(torch.randn(1, n_patches + 1, embed_dim))
        self.pos_drop = nn.Dropout(dropout)

        # Transformer encoder
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embed_dim,
            nhead=num_heads,
            dim_feedforward=int(embed_dim * mlp_ratio),
            dropout=dropout,
            activation='gelu',
            batch_first=True,
            norm_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=depth)

        # Classification head
        self.norm = nn.LayerNorm(embed_dim)
        self.head = nn.Sequential(
            nn.Linear(embed_dim, embed_dim // 2),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(embed_dim // 2, num_classes)
        )

    def forward(self, x):
        # x: (batch, 1, n_mels, time_steps)
        batch_size = x.shape[0]

        # Patch embedding
        x = self.patch_embed(x)  # (batch, n_patches, embed_dim)

        # Add CLS token
        cls_tokens = repeat(self.cls_token, '1 1 d -> b 1 d', b=batch_size)
        x = torch.cat([cls_tokens, x], dim=1)

        # Add position embedding
        x = x + self.pos_embed
        x = self.pos_drop(x)

        # Transformer
        x = self.transformer(x)

        # Classification from CLS token
        x = self.norm(x[:, 0])
        logits = self.head(x)

        return logits


# Créer le modèle
model = AudioTransformer(
    n_mels=N_MELS,
    time_steps=157,  # Calculé: (5*16000) // 512 + 1
    patch_size=16,
    embed_dim=256,
    depth=6,
    num_heads=8,
    num_classes=2,
    dropout=0.1
).to(DEVICE)

# Nombre de paramètres
total_params = sum(p.numel() for p in model.parameters())
print(f"🎯 Audio Transformer créé")
print(f"   Paramètres: {total_params:,}")
print(f"   Device: {DEVICE}")

# Test
with torch.no_grad():
    dummy_input = torch.randn(2, 1, 128, 157).to(DEVICE)
    dummy_output = model(dummy_input)
    print(f"   Test shape: {dummy_input.shape} → {dummy_output.shape}")
```

```python
class S4Layer(nn.Module):
    """Simplified State Space Model (S4) layer"""

    def __init__(self, d_model, state_dim=64):
        super().__init__()
        self.d_model = d_model
        self.state_dim = state_dim

        # Paramètres du SSM
        self.A = nn.Parameter(torch.randn(d_model, state_dim))
        self.B = nn.Parameter(torch.randn(d_model, state_dim))
        self.C = nn.Parameter(torch.randn(d_model, state_dim))
        self.D = nn.Parameter(torch.randn(d_model))

        # Normalisation
        self.norm = nn.LayerNorm(d_model)

    def forward(self, x):
        # x: (batch, seq_len, d_model)
        batch, seq_len, d = x.shape

        # Initialiser l'état
        h = torch.zeros(batch, self.d_model, self.state_dim, device=x.device)

        outputs = []
        for t in range(seq_len):
            # Update state: h_t = A * h_{t-1} + B * x_t
            h = torch.einsum('bds,ds->bds', h, torch.tanh(self.A)) + \
                torch.einsum('bd,ds->bds', x[:, t], self.B)

            # Output: y_t = C * h_t + D * x_t
            y = torch.einsum('bds,ds->bd', h, self.C) + x[:, t] * self.D
            outputs.append(y)

        out = torch.stack(outputs, dim=1)
        return self.norm(out + x)  # Residual connection


class MambaBlock(nn.Module):
    """Mamba block avec SSM"""

    def __init__(self, d_model, state_dim=64, expand_factor=2):
        super().__init__()

        self.norm1 = nn.LayerNorm(d_model)
        self.ssm = S4Layer(d_model, state_dim)

        self.norm2 = nn.LayerNorm(d_model)
        self.mlp = nn.Sequential(
            nn.Linear(d_model, d_model * expand_factor),
            nn.GELU(),
            nn.Linear(d_model * expand_factor, d_model)
        )

    def forward(self, x):
        # SSM branch
        x = x + self.ssm(self.norm1(x))

        # MLP branch
        x = x + self.mlp(self.norm2(x))

        return x


class AudioMamba(nn.Module):
    """Audio Mamba pour classification"""

    def __init__(self, n_mels=128, time_steps=157, patch_size=16,
                 d_model=256, n_layers=4, state_dim=64, num_classes=2):
        super().__init__()

        # Patch embedding
        self.patch_embed = PatchEmbedding(n_mels, time_steps, patch_size, d_model)
        n_patches = self.patch_embed.n_patches

        # Position embedding
        self.pos_embed = nn.Parameter(torch.randn(1, n_patches, d_model))

        # Mamba blocks
        self.blocks = nn.ModuleList([
            MambaBlock(d_model, state_dim) for _ in range(n_layers)
        ])

        # Classification head
        self.norm = nn.LayerNorm(d_model)
        self.head = nn.Sequential(
            nn.Linear(d_model, d_model // 2),
            nn.GELU(),
            nn.Dropout(0.1),
            nn.Linear(d_model // 2, num_classes)
        )

    def forward(self, x):
        # Patch embedding
        x = self.patch_embed(x)

        # Add position embedding
        x = x + self.pos_embed

        # Mamba blocks
        for block in self.blocks:
            x = block(x)

        # Global average pooling
        x = x.mean(dim=1)

        # Classification
        x = self.norm(x)
        logits = self.head(x)

        return logits


# Créer le modèle Mamba
model_mamba = AudioMamba(
    n_mels=N_MELS,
    time_steps=157,
    patch_size=16,
    d_model=256,
    n_layers=4,
    state_dim=64,
    num_classes=2
).to(DEVICE)

total_params_mamba = sum(p.numel() for p in model_mamba.parameters())
print(f"🐍 Audio Mamba créé")
print(f"   Paramètres: {total_params_mamba:,}")

# Choisir le modèle à entraîner
print("\n⚙️ Choisissez votre modèle:")
print("   1. Audio Transformer (model)")
print("   2. Audio Mamba/SSM (model_mamba)")
print("\n👉 Par défaut: Audio Transformer")

# Vous pouvez changer ici
SELECTED_MODEL = model  # ou model_mamba
MODEL_NAME = "Audio Transformer"  # ou "Audio Mamba"

print(f"\n✅ Modèle sélectionné: {MODEL_NAME}")
```

```python
def train_epoch(model, loader, criterion, optimizer, device):
    model.train()
    total_loss = 0
    correct = 0
    total = 0

    pbar = tqdm(loader, desc='Training')
    for specs, labels in pbar:
        specs, labels = specs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(specs)
        loss = criterion(outputs, labels)

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

        pbar.set_postfix({'loss': f'{loss.item():.4f}', 'acc': f'{100.*correct/total:.2f}%'})

    return total_loss / len(loader), 100. * correct / total


def validate(model, loader, criterion, device):
    model.eval()
    total_loss = 0
    correct = 0
    total = 0
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for specs, labels in tqdm(loader, desc='Validation'):
            specs, labels = specs.to(device), labels.to(device)

            outputs = model(specs)
            loss = criterion(outputs, labels)

            total_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()

            all_preds.extend(predicted.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    return total_loss / len(loader), 100. * correct / total, all_preds, all_labels


# Configuration
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(SELECTED_MODEL.parameters(), lr=LEARNING_RATE, weight_decay=0.01)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)

history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
best_acc = 0

print(f"🚀 Entraînement du {MODEL_NAME}")
print("="*60)

for epoch in range(EPOCHS):
    print(f"\nEpoch {epoch+1}/{EPOCHS}")

    train_loss, train_acc = train_epoch(SELECTED_MODEL, train_loader, criterion, optimizer, DEVICE)
    val_loss, val_acc, _, _ = validate(SELECTED_MODEL, test_loader, criterion, DEVICE)
    scheduler.step()

    history['train_loss'].append(train_loss)
    history['train_acc'].append(train_acc)
    history['val_loss'].append(val_loss)
    history['val_acc'].append(val_acc)

    print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
    print(f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.2f}%")

    if val_acc > best_acc:
        best_acc = val_acc
        torch.save(SELECTED_MODEL.state_dict(), f'/content/best_{MODEL_NAME.replace(" ", "_")}.pth')
        print(f"✅ Meilleur modèle sauvegardé!")

print(f"\n🎉 Entraînement terminé! Meilleure accuracy: {best_acc:.2f}%")
```

```python
# ============================================================================
# FONCTION POUR TESTER UN FICHIER AUDIO INDIVIDUEL
# ============================================================================

def predict_audio(model, file_path, device):
    """
    Prédire si un fichier audio est Normal ou Anomaly
    """
    model.eval()

    # Charger l'audio
    audio, _ = librosa.load(file_path, sr=SAMPLE_RATE, duration=DURATION, mono=True)

    # Ajuster la longueur
    if len(audio) > MAX_LENGTH:
        audio = audio[:MAX_LENGTH]
    else:
        audio = np.pad(audio, (0, max(0, MAX_LENGTH - len(audio))))

    # Normalisation
    audio = audio / (np.max(np.abs(audio)) + 1e-8)

    # Créer le mel-spectrogramme
    mel_spec = librosa.feature.melspectrogram(
        y=audio,
        sr=SAMPLE_RATE,
        n_mels=N_MELS,
        n_fft=N_FFT,
        hop_length=HOP_LENGTH
    )

    mel_spec_db = librosa.power_to_db(mel_spec, ref=np.max)
    mel_spec_db = (mel_spec_db - mel_spec_db.mean()) / (mel_spec_db.std() + 1e-8)

    # Prédiction
    with torch.no_grad():
        spec_tensor = torch.FloatTensor(mel_spec_db).unsqueeze(0).unsqueeze(0).to(device)
        output = model(spec_tensor)
        probs = F.softmax(output, dim=1)
        pred = output.argmax(1).item()

    return pred, probs[0].cpu().numpy()


# ============================================================================
# TESTER SUR DES EXEMPLES
# ============================================================================

print("\n" + "="*60)
print("🔍 TEST SUR FICHIERS AUDIO INDIVIDUELS")
print("="*60 + "\n")

# Prendre quelques fichiers de test
test_samples = test_files[:10]  # Les 10 premiers

for i, file_path in enumerate(test_samples):
    # Prédire
    pred, probs = predict_audio(SELECTED_MODEL, file_path, DEVICE)

    # Obtenir la vraie étiquette
    true_label = test_labels[test_files.index(file_path)]

    # Résultats
    filename = os.path.basename(file_path)
    pred_class = "🟢 NORMAL" if pred == 0 else "🔴 ANOMALY"
    true_class = "Normal" if true_label == 0 else "Anomaly"
    confidence = probs[pred] * 100

    # Status (correct ou pas)
    status = "✅" if pred == true_label else "❌"

    print(f"{i+1}. {filename}")
    print(f"   {status} Prédiction: {pred_class}")
    print(f"   Vraie classe: {true_class}")
    print(f"   Confiance: {confidence:.1f}%")
    print(f"   Probabilités: Normal={probs[0]:.3f}, Anomaly={probs[1]:.3f}")
    print("-" * 60)
```
